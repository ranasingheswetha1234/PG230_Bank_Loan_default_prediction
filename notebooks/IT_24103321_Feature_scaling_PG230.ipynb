**Task 4: Feature Scaling**

```python
# Import libraries used across the notebook
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler
from imblearn.over_sampling import SMOTE
```

```python
# Load dataset
df_en = pd.read_csv('C:\\Users\\User\\Desktop\\PG230\\PG230_Bank_Loan_default_prediction\\results\\outputs\\encoded_data.csv')
```

```python
# Numeric columns
num_cols = ['person_age','person_income','person_emp_exp',
            'loan_amnt','loan_int_rate','loan_percent_income',
            'cb_person_cred_hist_length','credit_score']
```

```python
# Histograms before scaling
df_en[num_cols].hist(figsize=(12,8))
plt.suptitle("Numeric Features Before Scaling")
plt.show()
```



* StandardScaling ensures that all features are scaled to equal ranges. It centers the data around 0 and scales by standard deviation, preserving the feature's distribution.



```python
# Apply StandardScaler
scaler = StandardScaler()
df_en[num_cols] = scaler.fit_transform(df_en[num_cols])
```

```python
# Histograms after scaling
df_en[num_cols].hist(figsize=(12,8))
plt.suptitle("Numeric Features After Scaling")
plt.show()
```

```python
# Create a new DataFrame with all features scaled in to equal range
df_en.to_csv('C:/Users/User/Desktop/PG230/PG230_Bank_Loan_default_prediction/results/outputs/Scaled_data.csv', index=False)
```
